```python
# QUESTION 2: ELECTRICITY BILL CALCULATION

# Write a Python program to calculate an electricity bill based on the
# number of units consumed.

# Conditions: - If units are less than or equal to 100, charge Rs. 2 per
# unit. - If units are greater than 100 and less than or equal to 200,
# charge Rs. 3 per unit. - If units are greater than 200, charge Rs. 5 per
# unit.

# Display the total electricity bill.

# Example: Enter electricity units: 150 Total Electricity Bill: Rs. 450

# Note: Apply the applicable rate to all units consumed, rather than
# slab-wise billing.

units = input()
units = int(units)

if units <= 100:
    bill = units * 2
elif units > 100 and units <= 200:
    bill = units * 3
else: 
    bill = units * 5

print(f"Total Elecctricity Bill: Rs. {bill}")
```

```python
# QUESTION 3: ATM WITHDRAWAL SYSTEM

# Write a Python program to simulate a simple ATM withdrawal.

# Conditions: - Initial balance = Rs. 10,000. - Get the withdrawal amount
# from the user. - Check whether the withdrawal amount is greater than
# zero. - Check whether sufficient balance is available. - If sufficient
# balance is available, deduct the amount and display the remaining
# balance. - Otherwise, display an insufficient balance message.

# Use nested if and else statements.

balance = 10000

amount = float(input(""))

if amount > 0:
    if amount <= balance:
        balance = balance - amount
        print("Withdrawal successful!")
        print("Remaining balance: Rs.", balance)
    else:
        print("Insufficient balance!")
else:
    print("Invalid amount! Withdrawal amount must be greater than zero.")
```

```python
# QUESTION 2: LIST MODIFICATION

# Write a Python program using the following list:

# numbers = [10, 20, 30, 40, 50]

# Perform the following operations: - Change the first element to 100. -
# Insert 25 at index 2. - Add 60 at the end of the list. - Remove the
# value 40. - Extend the list with [70, 80]. - Print the final list.

numbers = [10, 20, 30, 40, 50]

numbers[0] = 100          # Change the first element to 100
numbers.insert(2, 25)     # Insert 25 at index 2
numbers.append(60)        # Add 60 at the end
numbers.remove(40)        # Remove the value 40
numbers.extend([70, 80])  # Extend the list with [70, 80]

print("Final list:", numbers)

```

```python
# QUESTION 3: LIST OPERATIONS AND CALCULATIONS

# Write a Python program using the following list:

# marks = [85, 60, 95, 70, 80]

# Perform the following operations: - Sort the list in ascending order. -
# Sort the list in descending order. - Find the maximum mark. - Find the
# minimum mark. - Calculate the total marks. - Calculate the average
# marks. - Check whether all marks are greater than zero using all().

marks = [85, 60, 95, 70, 80]

ascending = sorted(marks)
descending = sorted(marks, reverse=True)

print("Ascending order:", ascending)
print("Descending order:", descending)
print("Maximum mark:", max(marks))
print("Minimum mark:", min(marks))

total = sum(marks)
average = total / len(marks)

print("Total marks:", total)
print("Average marks:", average)

print("All marks greater than zero:", all(m > 0 for m in marks))
```
