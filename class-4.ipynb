```python

# helps to check the quality of list 
# all() -> AND (all value should be non falsy value) true or false
# any() -> OR (atleast 1 non-falsy value shpuld be avaialble) true or false

# Falsy value: 0,"",None,[],(),False

list1 = [0,None,False,"",10]

print(all(list1))
print(any(list1))

list1 = [10,20,30,[]]
print(any(list1))
print(all(list1))

print(len(list1))
```
